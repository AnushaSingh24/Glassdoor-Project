## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# ============================================================
# DATA WRANGLING (FINAL ERROR FREE VERSION)
# ============================================================

# Create dataset copy
data = df.copy()

# Display shape before cleaning
print("Shape Before Cleaning :", data.shape)

# ============================================================
# DROP UNNECESSARY COLUMN
# ============================================================

if 'Unnamed: 0' in data.columns:
    data.drop('Unnamed: 0', axis=1, inplace=True)

# ============================================================
# REMOVE INVALID SALARY VALUES
# ============================================================

data = data[data['Salary Estimate'] != '-1']

# ============================================================
# CLEAN SALARY COLUMN
# ============================================================

# Convert to string
data['Salary Estimate'] = data['Salary Estimate'].astype(str)

# Remove all unwanted text
data['Salary Estimate'] = data['Salary Estimate'].str.replace('Employer Provided Salary:', '', regex=False)
data['Salary Estimate'] = data['Salary Estimate'].str.replace('Glassdoor est.', '', regex=False)
data['Salary Estimate'] = data['Salary Estimate'].str.replace('Employer est.', '', regex=False)
data['Salary Estimate'] = data['Salary Estimate'].str.replace('Per Hour', '', regex=False)

# Remove special symbols
data['Salary Estimate'] = data['Salary Estimate'].str.replace('$', '', regex=False)
data['Salary Estimate'] = data['Salary Estimate'].str.replace('K', '', regex=False)
data['Salary Estimate'] = data['Salary Estimate'].str.replace('(', '', regex=False)
data['Salary Estimate'] = data['Salary Estimate'].str.replace(')', '', regex=False)

# Remove extra spaces
data['Salary Estimate'] = data['Salary Estimate'].str.strip()

# ============================================================
# KEEP ONLY VALID SALARY RANGES
# ============================================================

data = data[data['Salary Estimate'].str.contains('-')]

# ============================================================
# SPLIT SALARY RANGE
# ============================================================

# Split minimum and maximum salary
salary_range = data['Salary Estimate'].str.split('-', expand=True)

# Create Min Salary column
data['Min Salary'] = salary_range[0].astype(int)

# Create Max Salary column
data['Max Salary'] = salary_range[1].astype(int)

# ============================================================
# CREATE AVERAGE SALARY
# ============================================================

data['Average Salary'] = (
    data['Min Salary'] + data['Max Salary']
) / 2

# ============================================================
# CREATE STATE COLUMN
# ============================================================

data['State'] = data['Location'].apply(
    lambda x: x.split(',')[-1].strip()
)

# ============================================================
# REMOVE NULL VALUES
# ============================================================

data.dropna(inplace=True)

# ============================================================
# REMOVE DUPLICATES
# ============================================================

data.drop_duplicates(inplace=True)

# ============================================================
# RESET INDEX
# ============================================================

data.reset_index(drop=True, inplace=True)

# ============================================================
# FINAL OUTPUT
# ============================================================

print("Shape After Cleaning :", data.shape)

# Display cleaned dataset
data.head()
````

### What all manipulations have you done and insights you found?

During the data wrangling process, the dataset was cleaned and prepared for analysis by handling missing values, removing duplicate records, and dropping unnecessary columns. The Salary Estimate column was processed by removing unwanted text and symbols, and new columns such as Minimum Salary, Maximum Salary, Average Salary, and State were created for better understanding of salary trends. Invalid salary records were also removed to improve data consistency. From the preprocessing stage, it was observed that salary varies significantly based on job role, company size, industry, and location. These manipulations improved the overall quality of the dataset and made it suitable for visualization, hypothesis testing, and machine learning model implementation.

````python

# CHART - 6 VISUALIZATION CODE
# TOP INDUSTRIES BY AVERAGE SALARY


# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
url = 'https://raw.githubusercontent.com/AnushaSingh24/Glassdoor-Project/refs/heads/main/glassdoor_jobs.csv'

df = pd.read_csv(url)

# Remove invalid salary rows
df = df[df['Salary Estimate'] != '-1']

# Clean salary column
df['Salary Estimate'] = df['Salary Estimate'].astype(str)

df['Salary Estimate'] = df['Salary Estimate'].str.replace('Glassdoor est.', '', regex=False)
df['Salary Estimate'] = df['Salary Estimate'].str.replace('Employer est.', '', regex=False)
df['Salary Estimate'] = df['Salary Estimate'].str.replace('Employer Provided Salary:', '', regex=False)

df['Salary Estimate'] = df['Salary Estimate'].str.replace('$', '', regex=False)
df['Salary Estimate'] = df['Salary Estimate'].str.replace('K', '', regex=False)
df['Salary Estimate'] = df['Salary Estimate'].str.replace('(', '', regex=False)
df['Salary Estimate'] = df['Salary Estimate'].str.replace(')', '', regex=False)

# Keep only salary ranges
df = df[df['Salary Estimate'].str.contains('-')]

# Split salary column
salary = df['Salary Estimate'].str.split('-', expand=True)

# Convert salary columns into numeric
df['Min Salary'] = pd.to_numeric(salary[0], errors='coerce')
df['Max Salary'] = pd.to_numeric(salary[1], errors='coerce')

# Remove null values
df.dropna(subset=['Min Salary', 'Max Salary'], inplace=True)

# Create Average Salary column
df['Average Salary'] = (df['Min Salary'] + df['Max Salary']) / 2

# Group industries by average salary
top_industries = df.groupby('Industry')['Average Salary'].mean().sort_values(ascending=False).head(10)

# Set figure size
plt.figure(figsize=(12,6))

# Create barplot
sns.barplot(
    x=top_industries.values,
    y=top_industries.index,
    palette='magma'
)

# Add chart title
plt.title('Top Industries by Average Salary', fontsize=16)

# Add labels
plt.xlabel('Average Salary')
plt.ylabel('Industry')

# Show chart
plt.show()
````

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.



*   Null Hypothesis (H₀): There is no significant relationship between company rating and average salary.
*   Alternate Hypothesis (H₁): There is a significant relationship between company rating and average salary.



#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

# COMPANY RATING VS AVERAGE SALARY

# Import library
from scipy.stats import pearsonr

# Remove null values
test_data = df[['Rating', 'Average Salary']].dropna()

# Perform Pearson Correlation Test
correlation, p_value = pearsonr(
    test_data['Rating'],
    test_data['Average Salary']
)

# Display results
print("Correlation Value :", correlation)
print("P-Value :", p_value)
````

##### Which statistical test have you done to obtain P-Value?

Pearson Correlation Test was used to obtain the p-value.

##### Why did you choose the specific statistical test?

Pearson Correlation Test is suitable because both variables (Company Rating and Average Salary) are numerical variables. It helps measure the strength and direction of the relationship between them.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.




*   Null Hypothesis (H₀): Average salary does not differ significantly across company sizes.
*   Alternate Hypothesis (H₁): Average salary differs significantly across company sizes.



#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

# COMPANY SIZE VS AVERAGE SALARY

# Import library
from scipy.stats import f_oneway

# Create salary groups
salary_groups = []

# Loop through company sizes
for size in df['Size'].dropna().unique():

    group = df[df['Size'] == size]['Average Salary']

    salary_groups.append(group)

# Perform ANOVA Test
f_statistic, p_value = f_oneway(*salary_groups)

# Display results
print("F-Statistic :", f_statistic)
print("P-Value :", p_value)
````

##### Which statistical test have you done to obtain P-Value?

One-Way ANOVA Test was used to obtain the p-value.

##### Why did you choose the specific statistical test?

ANOVA is appropriate because it compares the mean salary across multiple company size categories to determine whether significant differences exist.

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.




*   Null Hypothesis (H₀): There is no significant difference in average salary between different industries.
*   Alternate Hypothesis (H₁): There is a significant difference in average salary between different industries.



#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

# INDUSTRY VS AVERAGE SALARY

# Import library
from scipy.stats import f_oneway

# Select top industries
top_industries = df['Industry'].value_counts().head(5).index

# Create salary groups
industry_salary_groups = []

# Loop through industries
for industry in top_industries:

    group = df[df['Industry'] == industry]['Average Salary']

    industry_salary_groups.append(group)

# Perform One-Way ANOVA Test
f_statistic, p_value = f_oneway(*industry_salary_groups)

# Display results
print("F-Statistic :", f_statistic)
print("P-Value :", p_value)
````

##### Which statistical test have you done to obtain P-Value?

One-Way ANOVA Test was used to obtain the p-value.

##### Why did you choose the specific statistical test?

The ANOVA test is appropriate because it compares the mean average salary across multiple industries to determine whether significant salary differences exist among them.

## ***6. Feature Engineering & Data Pre-processing***

### 1. Handling Missing Values

````python
# Handling Missing Values & Missing Value Imputation

df['Rating'] = df['Rating'].fillna(df['Rating'].median())

df['Industry'] = df['Industry'].fillna(df['Industry'].mode()[0])

df['Sector'] = df['Sector'].fillna(df['Sector'].mode()[0])

df['Revenue'] = df['Revenue'].fillna(df['Revenue'].mode()[0])

df['Competitors'] = df['Competitors'].fillna('Unknown')

df.isnull().sum()
````

#### What all missing value imputation techniques have you used and why did you use those techniques?

Median imputation was used for numerical columns because it handles outliers effectively and prevents skewing of data. Mode imputation was used for categorical columns since it replaces missing values with the most frequent category. Constant value imputation was used for the competitors column by replacing missing values with “Unknown” to maintain data consistency.

### 2. Handling Outliers

````python
# Handling Outliers & Outlier Treatments

Q1 = df['Average Salary'].quantile(0.25)

Q3 = df['Average Salary'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR

upper_limit = Q3 + 1.5 * IQR

df = df[(df['Average Salary'] >= lower_limit) & (df['Average Salary'] <= upper_limit)]

print(df.shape)
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

IQR (Interquartile Range) method was used to detect and remove outliers because it is effective for skewed salary distributions and helps improve model performance by reducing extreme salary values.

### 3. Categorical Encoding

````python
# Encode Your Categorical Columns

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

categorical_columns = ['Industry', 'Sector', 'Size']

for col in categorical_columns:
    df[col] = le.fit_transform(df[col].astype(str))

df.head()
````

#### What all categorical encoding techniques have you used & why did you use those techniques?

Label Encoding was used to convert categorical text values into numerical form because machine learning models require numerical input data. It is simple and efficient for handling categorical variables.

#### 4. Removing URLs & Removing words and digits contain digits.

````python
# Remove URLs & Remove Words and Digits Containing Digits

import re

df['Job Description'] = df['Job Description'].apply(lambda x: re.sub(r'http\S+|www\S+|https\S+', '', x))

df['Job Description'] = df['Job Description'].apply(lambda x: re.sub(r'\w*\d\w*', '', x))

df['Job Description'].head()
````

#### 5. Removing Stopwords & Removing White spaces

````python
# Remove Stopwords

from nltk.corpus import stopwords

import nltk

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

df['Job Description'] = df['Job Description'].apply(
    lambda x: ' '.join([word for word in x.split() if word not in stop_words])
)

df['Job Description'].head()
````

````python
# Remove White Spaces

df['Job Description'] = df['Job Description'].str.strip()

df['Job Description'].head()
````

````python
# Normalizing Text (i.e., Stemming, Lemmatization etc.)
# Normalizing Text Using Lemmatization

from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

df['Lemmatized_Text'] = df['Tokens'].apply(
    lambda tokens: [lemmatizer.lemmatize(word) for word in tokens]
)

df[['Tokens', 'Lemmatized_Text']].head()
````

Lemmatization was used for text normalization because it converts words into their meaningful root forms while preserving context. It improves text quality and helps machine learning models understand similar words effectively.
